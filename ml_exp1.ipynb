```python
import pandas as pd
import numpy as np

df = pd.read_csv("student_exam_data.csv")
print("\n first 10")
print(df.head(10))

print("\n last 10")
print(df.tail(10))

print("\n Random 7 rows")
print(df.sample(7))

print(df.info())
print("\n missed val")
print(df.isnull().sum())

print("\n count and print total no of duplicate rows in dataframe")
print(df.duplicated().sum())

print(df.describe())

print("\n Access rows by label or position(21)")
print(df.loc[21])

print("\nShape of dataset:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("Rows and columns:", df.shape)
print(df.columns)
print("First 5 rows:")
print(df.head())

numeric = df.select_dtypes(include=np.number)

print("\nMaximum:")
print(np.max(numeric))
print("\nMinimum:")
print(np.min(numeric))
print("\nMean:")
print(np.mean(numeric))

df.sort_values(by='Previous Exam Score')
df.sort_values(by='Previous Exam Score', ascending=False)
```
